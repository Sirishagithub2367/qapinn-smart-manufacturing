```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

measurement_results = pd.DataFrame({

    "Measurement": [
        "Pauli-X",
        "Pauli-Z"
    ],

    "Qubits": [
        4,
        4
    ],

    "PDE Residual": [
        0.00258,
        0.007128
    ],

    "Relative L2 Error": [
        0.121416,
        0.13213
    ],

    "Training Time (s)": [
        600.591628,
        578.631692
    ],

    "Memory (MB)": [
        733.15625,
        673.175781
    ],

    "Parameters": [
        577,
        577
    ]

})

print(measurement_results)
```

```python
measurement_results.to_csv(
    "Measurement_Comparison.csv",
    index=False
)

print("Saved Successfully!")
```

PDE Residual Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    measurement_results["Measurement"],
    measurement_results["PDE Residual"],
    marker="o"
)

plt.xlabel("Measurement Operator")
plt.ylabel("PDE Residual")
plt.title("Measurement Operator vs PDE Residual")
plt.grid(True)

plt.show()
```

Relative L2 Error Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    measurement_results["Measurement"],
    measurement_results["Relative L2 Error"],
    marker="o"
)

plt.xlabel("Measurement Operator")
plt.ylabel("Relative L2 Error")
plt.title("Measurement Operator vs Relative L2 Error")
plt.grid(True)

plt.show()
```

Training Time Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    measurement_results["Measurement"],
    measurement_results["Training Time (s)"],
    marker="o"
)

plt.xlabel("Measurement Operator")
plt.ylabel("Training Time (s)")
plt.title("Measurement Operator vs Training Time")
plt.grid(True)

plt.show()
```

Memory Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    measurement_results["Measurement"],
    measurement_results["Memory (MB)"],
    marker="o"
)

plt.xlabel("Measurement Operator")
plt.ylabel("Memory (MB)")
plt.title("Measurement Operator vs Memory Usage")
plt.grid(True)

plt.show()
```

Parameters Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    measurement_results["Measurement"],
    measurement_results["Parameters"],
    marker="o"
)

plt.xlabel("Measurement Operator")
plt.ylabel("Trainable Parameters")
plt.title("Measurement Operator vs Trainable Parameters")
plt.grid(True)

plt.show()
```

```python
import pandas as pd

depth_results = pd.DataFrame({
    "Circuit Depth": [1, 2, 4],
    "PDE Residual": [0.004963, 0.007128, 0.006157],
    "Relative L2 Error": [0.158114, 0.132130, 0.217574],
    "Training Time (s)": [261.16, 578.63, 817.33],
    "Memory (MB)": [658.57, 673.18, 760.09],
    "Parameters": [565, 577, 601]
})

print(depth_results)
```

```python
depth_results.to_csv(
    "Circuit_Depth_Comparison.csv",
    index=False
)

print("Comparison table saved successfully!")
```

```python
print(depth_results.columns)
```

```python
print("depth_results" in globals())
print("plt" in globals())
```

Plot PDE Residual

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

plt.plot(
    depth_results["Circuit Depth"],
    depth_results["PDE Residual"],
    marker="o",
    linewidth=2
)

plt.xlabel("Circuit Depth")
plt.ylabel("PDE Residual")
plt.title("Circuit Depth vs PDE Residual")
plt.grid(True)

plt.show()
```

Plot Relative L2 Error

```python
plt.figure(figsize=(6,4))

plt.plot(
    depth_results["Circuit Depth"],
    depth_results["Relative L2 Error"],
    marker="o"
)

plt.xlabel("Circuit Depth")
plt.ylabel("Relative L2 Error")
plt.title("Circuit Depth vs Relative L2 Error")
plt.grid(True)

plt.show()
```

Plot Training Time

```python
plt.figure(figsize=(6,4))

plt.plot(
    depth_results["Circuit Depth"],
    depth_results["Training Time (s)"],
    marker="o"
)

plt.xlabel("Circuit Depth")
plt.ylabel("Training Time (seconds)")
plt.title("Circuit Depth vs Training Time")
plt.grid(True)

plt.show()
```

Plot Memory Usage

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

plt.plot(
    depth_results["Circuit Depth"],
    depth_results["Memory (MB)"],
    marker="o",
    linewidth=2
)

plt.xlabel("Circuit Depth")
plt.ylabel("Memory (MB)")
plt.title("Circuit Depth vs Memory Usage")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

plt.plot(
    depth_results["Circuit Depth"],
    depth_results["Parameters"],
    marker="o",
    linewidth=2
)

plt.xlabel("Circuit Depth")
plt.ylabel("Trainable Parameters")
plt.title("Circuit Depth vs Trainable Parameters")
plt.grid(True)

plt.show()
```

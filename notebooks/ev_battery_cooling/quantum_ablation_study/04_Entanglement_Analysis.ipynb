```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

ent_results = pd.DataFrame({
    "Entanglement": ["Ring", "Linear", "None"],
    "Qubits": [4, 4, 4],
    "PDE Residual": [0.010745, 0.006352, 0.004181],
    "Relative L2 Error": [0.211139, 0.219389, 0.146274],
    "Training Time (s)": [504.23, 571.93, 725.92],
    "Memory (MB)": [671.54, 709.93, 831.29],
    "Parameters": [577, 577, 577]
})

print(ent_results)
```

```python
ent_results.to_csv(
    "Entanglement_Comparison.csv",
    index=False
)

print("Saved Successfully!")
```

PDE Residual Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    ent_results["Entanglement"],
    ent_results["PDE Residual"],
    marker="o"
)

plt.xlabel("Entanglement")
plt.ylabel("PDE Residual")
plt.title("Entanglement vs PDE Residual")
plt.grid(True)

plt.show()
```

Relative L2 Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    ent_results["Entanglement"],
    ent_results["Relative L2 Error"],
    marker="o"
)

plt.xlabel("Entanglement")
plt.ylabel("Relative L2 Error")
plt.title("Entanglement vs Relative L2 Error")
plt.grid(True)

plt.show()
```

Training Time Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    ent_results["Entanglement"],
    ent_results["Training Time (s)"],
    marker="o"
)

plt.xlabel("Entanglement")
plt.ylabel("Training Time (s)")
plt.title("Entanglement vs Training Time")
plt.grid(True)

plt.show()
```

Memory Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    ent_results["Entanglement"],
    ent_results["Memory (MB)"],
    marker="o"
)

plt.xlabel("Entanglement")
plt.ylabel("Memory (MB)")
plt.title("Entanglement vs Memory Usage")
plt.grid(True)

plt.show()
```

Parameters Plot

```python
plt.figure(figsize=(6,4))

plt.plot(
    ent_results["Entanglement"],
    ent_results["Parameters"],
    marker="o"
)

plt.xlabel("Entanglement")
plt.ylabel("Parameters")
plt.title("Entanglement vs Trainable Parameters")
plt.grid(True)

plt.show()
```
